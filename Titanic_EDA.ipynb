```python
# ==========================================
# CodeAlpha Data Analytics Internship
# Task 2: Exploratory Data Analysis (EDA)
# Dataset: Titanic
# Author: [Your Name]
# ==========================================

# Step 1: Import all required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure plots for better display
%matplotlib inline
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (8, 5)

print("=" * 55)
print("   TITANIC DATASET - EXPLORATORY DATA ANALYSIS")
print("=" * 55)

# ==========================================
# Step 2: Load the Dataset
# ==========================================
df = sns.load_dataset('titanic')

print("\n--- First 5 Rows of Dataset ---")
print(df.head())

# ==========================================
# Step 3: Explore Data Structure
# ==========================================
print("\n--- Dataset Information ---")
print(df.info())

print("\n--- Statistical Summary ---")
print(df.describe())

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# ==========================================
# Step 4: Define Research Questions
# ==========================================
print("\n" + "=" * 55)
print("RESEARCH QUESTIONS:")
print("Q1: How does survival vary by gender?")
print("Q2: How does survival vary by passenger class?")
print("Q3: What is the age distribution of passengers?")
print("Q4: Which factors are most correlated with survival?")
print("=" * 55)

# ==========================================
# Step 5: Visualization 1 - Age Distribution
# ==========================================
plt.figure(figsize=(8, 5))
sns.histplot(df['age'].dropna(), kde=True, bins=30, color='skyblue')
plt.title('Age Distribution of Passengers', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# ==========================================
# Step 6: Visualization 2 - Passenger Class Count
# ==========================================
plt.figure(figsize=(6, 4))
sns.countplot(x='pclass', data=df, palette='viridis')
plt.title('Passenger Class Count', fontsize=14)
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()

# ==========================================
# Step 7: Visualization 3 - Survival by Gender
# ==========================================
plt.figure(figsize=(6, 4))
sns.countplot(x='sex', hue='survived', data=df, palette='Set2')
plt.title('Survival by Gender', fontsize=14)
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

# ==========================================
# Step 8: Visualization 4 - Survival by Passenger Class
# ==========================================
plt.figure(figsize=(6, 4))
sns.countplot(x='pclass', hue='survived', data=df, palette='Set1')
plt.title('Survival by Passenger Class', fontsize=14)
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

# ==========================================
# Step 9: Visualization 5 - Correlation Heatmap
# ==========================================
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix', fontsize=14)
plt.show()

# ==========================================
# Step 10: Key Findings and Conclusion
# ==========================================
print("\n" + "=" * 55)
print("KEY FINDINGS:")
print("=" * 55)
print("1. Females had a significantly higher survival rate than males.")
print("2. First-class passengers survived more than third-class passengers.")
print("3. Third-class passengers had the lowest survival rate.")
print("4. 'Fare' and 'Pclass' show strong correlation with survival.")
print("5. 'Age' column contains 177 missing values (Data Issue).")
print("=" * 55)
print("EDA COMPLETED SUCCESSFULLY!")
print("=" * 55)
```

```python
# ==========================================
# Step 11: Data Cleaning (Bonus)
# ==========================================
print("\n--- Data Cleaning: Filling Missing Age Values ---")
df['age'].fillna(df['age'].median(), inplace=True)
print("Missing values in 'age' after cleaning:", df['age'].isnull().sum())

# Verify no missing values remain in age
print("\n--- Updated Missing Values ---")
print(df.isnull().sum())
```
